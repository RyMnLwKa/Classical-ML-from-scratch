# **AUTOENCODER**

```python
class WrongInitType(Exception):
  pass

class Dense(tf.Module):

  def __init__(self, output, activation_method = None, init = "normal", bias_use = True):
    super().__init__(name = f"{activation_method.__name__}Dense")
    self.output = output
    self.act = activation_method
    self.init = init
    self.bias_use = bias_use

  def init_weights(self, input, output, init):

    if isinstance(init, (np.ndarray, tf.Tensor)):
      return tf.cast(init, dtype = tf.float32)

    if init == "normal":
      mean, stddev = 0, (2 / (input + output))**0.5
      return tf.random.normal([input, output], mean = mean, stddev = stddev)
    elif init == "uniform":
      b = (6.0 / (input + output)) ** 0.5
      return tf.random.uniform([input, output], minval=-b, maxval=b)
    else:
      raise WrongInitType("Expected 'uniform', 'normal' or or a custom matrix (np.ndarray / tf.Tensor)")

  def __call__(self, x):
    if not hasattr(self, 'w'): # saves from random init for each __call__ instead of using trained weights
      self.w = tf.Variable(self.init_weights(x.shape[-1], self.output, self.init))
      self.b = tf.Variable(tf.zeros([self.output])) if self.bias_use else 0
    y = x @ self.w + self.b
    return self.act(y) if self.act is not None else y

def linear(y):
  return y

def relu(y):
  return tf.maximum(0.0, y)

def tf_relu(y):
  return tf.nn.relu(y)

def sigmoid_infcut(y):
  row_exp = tf.exp(-y)
  is_inf = tf.math.is_inf(row_exp)
  max_val = tf.float32.max
  valid_exp = tf.where(is_inf, max_val, row_exp)
  return 1 / (1 + valid_exp)

def sigmoid_valneg_risky(y):
  positive = 1.0 / (1.0 + tf.exp(-y)) #multiple and devide exp(y) and get formula for negative
  negative = tf.exp(y) / (1 + tf.exp(y)) #multiple and devide exp(-y) and get formula for positive (standart)
  exp = tf.where(y >=0, positive, negative)
  return exp

def sigmoid_valneg_slowly(y):
  def _single_sigmoid(val):
    return tf.cond(val >= 0,
                  lambda: 1.0 / (1.0 + tf.exp(-val)),
                  lambda: tf.exp(val) / (1.0 + tf.exp(val)))
  return tf.map_fn(_single_sigmoid, y)

def tf_sigmoid(y):
  return tf.nn.sigmoid(y)
```

Вот формула KL-дивергенции для j-го нейрона:

$$
D_{KL}(\rho \| \hat{\rho}_j) = \rho \log \frac{\rho}{\hat{\rho}_j} + (1 - \rho) \log \frac{1 - \rho}{1 - \hat{\rho}_j}
$$

И полная сумма по всем нейронам:

$$
\mathcal{L}_{KL} = \sum_{j=1}^{D} \left( \rho \log \frac{\rho}{\hat{\rho}_j} + (1 - \rho) \log \frac{1 - \rho}{1 - \hat{\rho}_j} \right)
$$

```python
class GeneralLoss( ):

  def __init__(self):
    self.penalties = []

  def _basic_loss(self, x, decoded_x):
    return tf.reduce_mean(tf.square(x - decoded_x)) #if _basic_loss uses as aggfunc sum - all penalties use sum, otherwise all uses mean (except KL)

  def add(self, method, weight, **kwargs):

    penalty = {'name' : method.__name__,
               'func' : method,
               'lambda': weight,
               'kwargs' : kwargs}

    self.penalties.append(penalty)

    return self

  def __call__(self, x, decoded_x, z):

    total_loss = 0.0
    total_loss += self._basic_loss(x, decoded_x)

    for pen in self.penalties:
      pen_value = pen['lambda'] * pen['func'](z, **pen['kwargs'])
      total_loss += pen_value

    return total_loss


def orthogonal_penalty(z, eps=1e-12, orthonormal = False):

    Z = z

    if not orthonormal:
      l2_norm = tf.norm(z, axis=0)
      normalised_z = z / tf.maximum(eps, l2_norm)  #by using normalization we terminate variance, thus training and loss dont know anything about scale - we can achieve only ORTHOGONALITY (like “covariance_eigh” )
      Z = normalised_z

    gram_matrix = tf.transpose(Z) @ Z             #but without normalization we force model to adapt to scale and angles during training, thus we obtain ORTHONORMALITY
    identity_matrix = tf.eye(tf.shape(Z)[-1])

    mse = tf.square(gram_matrix - identity_matrix)
    return tf.reduce_mean(mse)

def lasso_penalty(z):
  return tf.reduce_mean(tf.abs(z))

def ridge_penalty(z):
  return tf.reduce_mean(tf.square(z))

def kl_penalty(z, p=0.05, eps=1e-12):
  average_activations = tf.reduce_mean(z, axis=0)
  safe_average_activations = tf.clip_by_value(average_activations, eps, 1-eps)
  return tf.reduce_sum(p*tf.math.log(p/safe_average_activations) + (1-p)*tf.math.log((1-p)/(1-safe_average_activations)))

def monotonic_variance_penalty(z):
  variances = tf.math.reduce_variance(z, axis=0)
  diffs = variances[1:] - variances[:-1]
  penalty = tf.reduce_mean(tf.nn.relu(diffs))
  return penalty
```

# RULE (*how to train ortho-*):
- pca + whiten=False -> orthogonal features (samples still can be correlated and for all ones below)
- ae + normalisation = True-> orthogonal features (normalisation makes model blind for scale, thus we train only orthogonality
- pca + white = True -> orthonormal features
- ae + normalisation = False -> orthonormal features (model sees scale and angles, thereby it can be trained as orthonormal matrix
# CONCLUSION:
- pca + white = True === ae + normalisation = False;
- if we **ignore diagonal** (by normalisation of variance) model learns only angles (not diagonal elements) -> orthogonal matrix (**reach angles = 90**)

---



```python
class Autoencoder(tf.Module):
  def __init__(self, latent_size, transitional_size, output_size = None):
    super().__init__()
    self.latent_size = latent_size
    self.output_size = output_size

    self.input_enc = Dense(transitional_size, activation_method=relu, init="normal", bias_use=True)
    self.latent_enc = Dense(latent_size, activation_method=relu, init = "normal", bias_use=True)
    self.latent_dec = Dense(transitional_size, activation_method=relu, bias_use=True)

    if output_size is not None:
      self.output_dec = Dense(output_size, activation_method=linear, bias_use=True)

  def __call__(self, x):

    if not hasattr(self, 'output_dec'):
      self.output_dec = Dense(x.shape[-1], activation_method=linear)

    z = self.latent_enc(self.input_enc(x))
    decoded_x = self.output_dec(self.latent_dec(z))
    return z, decoded_x

  def fit(self, x, optimizer, loss_func, epochs=10):
    @tf.function
    def train_step(x):
      with tf.GradientTape() as tape:
        z, decoded_x = self(x)
        loss = loss_func(x, decoded_x, z)
      grads = tape.gradient(loss, self.trainable_variables)           # это сами градиенты
      optimizer.apply_gradients(zip(grads, self.trainable_variables)) # это шаг обновления w = w - lr*grad_w
      return loss

    loss_history = []

    for epoch in range(epochs):

      epoch_loss = 0.0
      num_batches = 0

      for batch_x in x: #batch_size set in advance in tf.data.Dataset
        loss = train_step(batch_x)
        epoch_loss += loss
        num_batches += 1

      mean_loss = epoch_loss / num_batches
      loss_history.append(mean_loss.numpy())
      print(f"Epoch {epoch+1}: {loss_history[-1]:.4f}")

    return loss_history

  def encode(self, x):
    return self.latent_enc(self.input_enc(x))

  def restore(self, x):
    z, decoded_x = self(x)
    return decoded_x
```

# **ОБУЧЕНИЕ**/**ТЕСТЫ**

```python
(X, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

X = X.reshape(-1, 784).astype('float32') / 255.0
x_test = x_test.reshape(-1, 784).astype('float32') / 255.0
print(f"X.shape: {tf.shape(X).numpy()}, x_test.shape: {tf.shape(x_test).numpy()}")

# optional centering
mean_x_train = tf.reduce_mean(X, axis=0)
centered_x_train = X - mean_x_train
centered_x_test = x_test - mean_x_train # we use mean_x_train (not mean_x_test) to avoid DATA LEAKAGE

batch_size = 256
train_dataset = (
    tf.data.Dataset.from_tensor_slices(centered_x_train)
    .shuffle(buffer_size=10000)
    .batch(batch_size)
    .prefetch(tf.data.AUTOTUNE)
)

test_dataset = (
    tf.data.Dataset.from_tensor_slices(centered_x_test)
    .batch(batch_size)
    .prefetch(tf.data.AUTOTUNE)
)


autoencoder = Autoencoder(latent_size=64, transitional_size = 128, output_size=784)

optimizer = tf.keras.optimizers.Adam(learning_rate=0.001)


loss_fn = (
    GeneralLoss()
    #.add(lasso_penalty, weight=0.001)
    #.add(ridge_penalty, weight=0.001)
    #.add(orthogonal_penalty, weight=0.01)
    #.add(monotonic_variance_penalty, weight=0.01)
    #.add(kl_penalty, weight=0.1)
)

print("Training session started...")

with tf.device('/CPU:0'):
  history = autoencoder.fit(
      x=train_dataset,
      optimizer=optimizer,
      loss_func=loss_fn,
      epochs=5
  )

print("<DONE>")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(history) + 1), history, marker='o', color='b', linestyle='-')
plt.title('Loss: autoencoder')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True)
plt.xticks(range(1, len(history) + 1))
plt.show()
```

### **Explanation for subspace_angles beetween latent basis**

**What are we measuring?**

`Subspace_angles` computes the principal angles between two subspaces (in this case, between the latent space bases of the AE and PCA). The subspaces are defined by the columns of the weight matrices. Pairs of columns and their corresponding (paired) axes from the AE and PCA bases are searched to find those that yield the smallest angle. If both models have learned the exact same subspace, the angles approach 0°. The larger the angle, the more the axis directions diverge, which allows us to numerically measure how equivalent the AE basis is to the PCA basis.

**Latent space basis in AE and PCA**

The weights `input_encoder.w @ latent_encoder.w` in the autoencoder serve as the analog of the eigenvectors in PCA, with the same dimension `DxK`.
Why? In PCA, projecting an `NxD` matrix onto them results in an `NxK` matrix, i.e., `K` independent components.
In the autoencoder, the `latent_encoder` layer performs an analogous projection through the activation function `(wx+b)`.

**Bias**

The autoencoder uses a bias, which is responsible for shifting the latent space away from the origin of the axes.
However, in PCA there is no shift; the data only changes direction, defined by the projection onto the eigenbases (`kx+b`, where `k` is the rotation coefficient). That is, the components pass through the origin.
Since we use `W`, we omit `b`, which also affects the latent space, so one could try to add it as the first column via `hstack`. But then, to match the dimensions, a zero column must be added to the `DxK` PCA matrix.

**Centering**

In PCA, centering not only defines the main task-finding the axes of maximum variance - but also removes the bias (shift). If there were no centering, one could introduce `b` to compensate for the shift and return the data to the origin (in order to describe the data variance).
How it works: by subtracting the mean, the arithmetic mean of the data vectors becomes 0, which shifts the data so that the center of the data cloud coincides with the origin of the original basis.
Can centering be applied to an AE? Yes, since bias serves the same function as centering (making the axes pass through zero in the original basis). Therefore, before training, it is necessary to center the data `X` (this does not affect PCA, as there will be no double centering because the second time the mean subtracted will simply be zero).

**Conclusion**

By centering the data, we force the autoencoder to behave like PCA, considering the data variance around a zero mean. However, this is not always beneficial, especially when working with images and sparse data, or under non-linearity (e.g., a sigmoid on the output during reconstruction may suffer from a distorted scale). Sometimes the bias should be non-zero, but in linear models, this is convenient. **But it only works when linear AE and PCA find close latent spaces**, that is possible only with initialising PCA components as AE encoders weights.

### **Initialization for Dense to find proper PCA latent space**

```python
import numpy as np

W_pca = pca.components_.T
n_components = z_pca.shape[-1]
transitional_size = 128

U, S, Vt = np.linalg.svd(W_pca, full_matrices=False)

U_aug = np.zeros((784, transitional_size))
U_aug[:, :n_components] = U

S_aug = np.zeros(transitional_size)
S_aug[:n_components] = S

Vt_aug = np.zeros((transitional_size, n_components))
Vt_aug[:n_components, :] = Vt

S_sqrt = np.sqrt(S_aug)

# W_input: (784, transitional_size (128))
W_input = U_aug * S_sqrt

# W_latent: (transitional_size (128), n_components (64))
W_latent = np.diag(S_sqrt) @ Vt_aug

W_reconstructed = W_input @ W_latent
print("Difference with PCA:", np.linalg.norm(W_pca - W_reconstructed))
```

# **CONCLUSION: Are linear AE and PCA just the same?**

- **1. Why AE and PCA diverge without initialization (60–70°):**
Under random initialization, the Autoencoder (AE) searches for *any* of the infinitely many 32-dimensional subspaces that minimize the reconstruction MSE (MAE ~0.05). It is indifferent to the variance distribution of the latents and their ordering. In contrast, PCA identifies a unique, strictly determined subspace where the axes are sorted by descending maximum variance. Both models find excellent solutions, but they look in different directions, hence the large principal angles.

- **2. Difference in variance distribution:**
In PCA, the axes are strictly ordered by descending variance (the first is the "most important," and the last is the "noisiest"). In the AE, the variance can be uniformly "smeared" across all 32 latents (or distorted by L1/Orthogonal penalties), because for reconstruction, the total information matters, not the hierarchy of features.

- **3. Final proof (2.5°):**
The fact that initializing the AE weights with PCA components (via SVD with zero-padding) drops the principal angle from 70° to **2.5°** proves that the **AE is mathematically capable of learning the exact same subspace as PCA**. The random gradient descent simply finds an alternative optimum not making an error. The residual 2.5° is merely numerical `float32` noise and the artifacts of the zero-padded columns during matrix multiplication.

- **4. Alternative proof via Gram Matrices:**
Comparing the Gram matrices (which capture the pairwise similarity of all data points) confirms this. With random initialization, the mean absolute difference between AE and PCA similarity matrices is **0.153** (max 0.676). After the PCA initialization, this difference drops by over 3 times to **0.046** (max 0.396). This proves that the AE not only aligns its axes but also fundamentally replicates closely PCA's data topology and pairwise relationships.

```python
X.info()

#X.select_dtypes(exclude = ['object']) # проверили: все object
for col_idx, col_name in enumerate(X.columns):
    try:
        #X.iloc[:, col_idx] = X.iloc[:,col_idx].astype(np.float32).values  # не работает, iloc|loc перезаписывают значения, но не тип
        X[col_name] = X[col_name].astype(np.float32)
    except (TypeError, ValueError) as e:
        try:
            X[col_name] = X[col_name].str.decode("UTF-8")  #декодируем байты в строки
        except (TypeError, ValueError) as e:
            pass

X.head()
```

```python
X.info()

is_col_has_nan = np.any(X.isna().to_numpy(), axis = 0).reshape(41) # проверка на at least one NaN
is_col_only_zero = np.all(X.to_numpy(), axis = 0).reshape(41) # проверка на чисто нулевые столбцы (True, если все ненулевые), поэтому ~
popential_drop = X.loc[:,is_col_only_zero]
is_col_str_or_zero = np.all(~popential_drop.fillna('').astype(bool).to_numpy(), axis=0) # или map(lambda x: bool(x)) работает поэлементно, а apply аггрегирует только по осям/таблице
is_col_only_zero[is_col_only_zero] = is_col_str_or_zero # послдние 4 строки проверяют столбцы с True из is_col__only_zero, потому что np выдает считает пустыми ячейки типа str/bytes
X = X.loc[:,~is_col_only_zero]              # поэтому необходимо убедиться это строки или реальные нули/None/''

X.head()
```

```python
import pandas as pd

X = pd.get_dummies(X) # one-hot для строковых
X = X.astype(np.float32)

X.info()
X.head()
```

```python
# TRAIN/VAL
train_z, val_z = AE.encode(x_train.to_numpy()), AE.encode(x_val.to_numpy())
train_restored_x, val_restored_x = AE.restore(x_train.to_numpy()), AE.restore(x_val.to_numpy())

#1. reconstruction error
error_threshold = None
sum_mae_per_object_train = np.mean(np.abs(x_train.to_numpy() - train_restored_x.numpy()), axis = 1)
mean_mae_per_object_train = np.mean(np.abs(x_train.to_numpy() - train_restored_x.numpy()), axis=1)                
mean_mae_train = np.mean(sum_mae_per_object_train)

sum_mae_per_object_val = np.sum(np.abs(x_val.to_numpy() - val_restored_x.numpy()), axis = 1)
mean_mae_per_object_val = np.mean(np.abs(x_val.to_numpy() - val_restored_x.numpy()), axis=1)                              #ЭТО ДЛЯ ПОРОГА
mean_mae_val = np.mean(sum_mae_per_object_val) 

# 2. latent_space + distance
dist_eu_threshold = None
dist_mh_threshold = None

mean_train_z = np.mean(train_z.numpy(), axis = 0)
diff = val_z.numpy()[:, np.newaxis, :] - mean_train_z[np.newaxis, np.newaxis, :]

# euclidian (metric = ...)
dist_to_centre_eu = np.sqrt(np.maximum(0,np.sum(diff ** 2, axis = 1)))
#dist_to_centre_eu = cdist(val_z, mean_train_z.reshape(1, -1), metric = "euclidean")
mean_dist_to_centre_eu_per_object = np.mean(dist_to_centre_eu ,axis = 1)                                 #ЭТО ДЛЯ ПОРОГА

# mahalanobis
X = np.vstack([val_z.numpy(), mean_train_z])
cov_X = np.cov(X, rowvar = False) #rowvar = False because correletion in columns not rows
inv_cov_X = np.linalg.pinv(cov_X)
dist_to_centre_mh = np.sqrt(np.maximum(0, np.sum(diff * (diff @ inv_cov_X), axis=1))) #np.matmul
#dist_to_centre_mh = cdist(val_z, mean_train_z.reshape(1, -1), metric = "mahalanobis", VI = inv_cov_X)
mean_dist_to_centre_mh_per_object = np.mean(dist_to_centre_mh ,axis = 1)                                                   #ЭТО ДЛЯ ПОРОГА

# 3. latent_space + one class classification
iforest = IsolationForest(n_estimators=200, random_state = 42, bootstrap=True, contamination = 0.05)
ocsvm = OneClassSVM(kernel = "rbf", nu = 0.05)

iforest.fit(train_z)
ocsvm.fit(train_z)

pred_iforest = iforest.predict(val_z)
pred_ocsvm = ocsvm.predict(val_z)

score_iforest = iforest.decision_function(val_z)
score_ocsvm = ocsvm.decision_function(val_z)

# 4. combination
alpha = 0.5
beta = 0.5

weighted_score = alpha * mean_mae_per_object_val + beta * mean_dist_to_centre_mh

# 5. vae
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import classification_report

metrics_data = {"Precision": [], "Recall": [], "F1-score": []}
model_names = []

for name, preds in predictions.items():
    report = classification_report(
        is_actual_anomaly, 
        preds, 
        target_names=["Normal", "Anomaly"], #аргумент target_names строго соответствует порядку уникальных меток. То есть [0,1], поэтому такой порядок
        output_dict=True,
        zero_division=0
    )
    
    model_names.append(name)
    metrics_data["Precision"].append(report["Anomaly"]["precision"])
    metrics_data["Recall"].append(report["Anomaly"]["recall"])
    metrics_data["F1-score"].append(report["Anomaly"]["f1-score"])


x = np.arange(len(model_names))
width = 0.25

fig, ax = plt.subplots(figsize=(12, 6))

rects1 = ax.bar(x - width, metrics_data["Precision"], width, label='Precision')
rects2 = ax.bar(x, metrics_data["Recall"], width, label='Recall')
rects3 = ax.bar(x + width, metrics_data["F1-score"], width, label='F1-score')

ax.set_ylabel('Score')
ax.set_title('Anomaly Detection Metrics by Methods')
ax.set_xticks(x)
ax.set_xticklabels(model_names, rotation=15, ha="right")
ax.legend(loc='lower right')
ax.set_ylim(0, 1.1) # метрики от 0 до 1

def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.2f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=9)

autolabel(rects1)
autolabel(rects2)
autolabel(rects3)

plt.tight_layout()
plt.show()
```

# **Conclusions:**

**1. Latent + Mahalanobis - The Absolute Winner**
Demonstrated the best and most stable performance. At the 95th percentile, it achieved a Precision of 0.98 and a Recall of 1.00 (F1 = 0.99). Mahalanobis distance outperforms Euclidean distance because it accounts for correlations between latent features, which is critical for complex high-dimensional data.

**2. Latent + OCSVM and Isolation Forest — Excellent Alternatives**
Both methods showed consistently high Recall (1.00), practically avoiding missed attacks (False Negatives). OCSVM is slightly more precise (Precision 0.97), whereas Isolation Forest requires less hyperparameter tuning.

**3. Reconstruction Error (MAE) — A Solid Baseline**
Showed surprisingly high Precision (0.97) and Recall (1.00) at the 97th percentile. However, in previous experiments (at the 99th percentile), MAE often missed attacks (low Recall) that the autoencoder "memorized" and reconstructed too well (maybe because a overfitting - a lot of epochs). Therefore, it is less reliable than Mahalanobis.

**4. Latent + Euclidean — Unstable**
At the 95th percentile, it showed excellent results (F1 = 0.97), but when shifted to the 97th percentile, its Recall dropped sharply to 0.87. This proves that Euclidean distance is sensitive to the scale and shape of the latent cloud, making its threshold harder to tune.

**Key Takeaways:**

*   **The $\text{Mean} + 3\sigma$ rule does not work.** Q-Q plots revealed heavy right tails in the error distributions. Therefore, percentiles (the 95th or 97th) must be used for threshold selection.
*   **Choosing a percentile is a business logic decision:**
    *   The **95th percentile** ensures maximum Recall (catches 100% of attacks) but yields slightly more False Positives on normal data (Precision ~0.97).
    *   The **97th percentile** provides near-perfect Precision (~0.99) but may miss a portion of the attacks.
*   **Autoencoder + One-Class classifiers (OCSVM, IF)** effectively solve the "curse of dimensionality" by allowing the algorithms to operate in a clean, compact latent feature space.
